```python
# =========================================================
# STAGE 5 — MACHINE LEARNING
# =========================================================

import pandas as pd
import numpy as np

import joblib

# Machine Learning
from sklearn.svm import LinearSVC

# Evaluation Metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

import warnings
warnings.filterwarnings("ignore")
```

```python
# =========================================================
# LOAD TRAIN TEST DATA
# =========================================================

X_train = joblib.load("X_train.pkl")
X_test = joblib.load("X_test.pkl")

y_train = joblib.load("y_train.pkl")
y_test = joblib.load("y_test.pkl")

print("Artifacts Loaded Successfully")

print()

print("X_train Shape:")
print(X_train.shape)

print()

print("X_test Shape:")
print(X_test.shape)
```

```python
# =========================================================
# INITIALIZE MODEL
# =========================================================

svm_model = LinearSVC(
    random_state=42
)

print("Linear SVM Initialized")
```

```python
# =========================================================
# TRAIN MODEL
# =========================================================

print("Training Linear SVM Model...")

svm_model.fit(
    X_train,
    y_train
)

print("Model Training Completed")
```

```python
# =========================================================
# PREDICTIONS
# =========================================================

y_pred = svm_model.predict(
    X_test
)

print("Predictions Generated")

print()

print("Sample Predictions:")
print(y_pred[:20])
```

```python
# =========================================================
# ACCURACY
# =========================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Accuracy:")
print(round(accuracy, 4))

# =========================================================
# PRECISION
# =========================================================

precision = precision_score(
    y_test,
    y_pred
)

print("Precision:")
print(round(precision, 4))

# =========================================================
# RECALL
# =========================================================

recall = recall_score(
    y_test,
    y_pred
)

print("Recall:")
print(round(recall, 4))

# =========================================================
# F1 SCORE
# =========================================================

f1 = f1_score(
    y_test,
    y_pred
)

print("F1 Score:")
print(round(f1, 4))

# =========================================================
# CLASSIFICATION REPORT
# =========================================================

report = classification_report(
    y_test,
    y_pred
)

print(report)

# =========================================================
# CONFUSION MATRIX
# =========================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("Confusion Matrix:")

print(cm)
```

```python
# =========================================================
# SAVE TRAINED MODEL
# =========================================================

joblib.dump(
    svm_model,
    "linear_svm_model.pkl"
)

print("Trained Model Saved Successfully")
```

```python
# =========================================================
# STAGE 5 SUMMARY
# =========================================================

print("=" * 60)

print("STAGE 5 COMPLETED SUCCESSFULLY")

print("=" * 60)

print()

print("Model Used:")
print("Linear SVM")

print()

print("Accuracy:")
print(round(accuracy, 4))

print()

print("Precision:")
print(round(precision, 4))

print()

print("Recall:")
print(round(recall, 4))

print()

print("F1 Score:")
print(round(f1, 4))
```
